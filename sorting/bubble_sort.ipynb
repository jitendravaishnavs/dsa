```python
nums =[5,8,1,6,9,2,4]

def bubble_sort(nums):
    n= len(nums)
    for i in range(0,n):
        for j in range(i+1,n):
            if nums[i]>nums[j]:
                nums[j],nums[i]= nums[i],nums[j]
    print(nums)
    
            


bubble_sort(nums)
```

```python
nums =[5,8,1,6,9,2,4]

def bubble_sort(nums):
    n= len(nums)
    for i in range(n-2,-1, -1):
        for j in range(0,i+1):
            if nums[j]>nums[j+1]:
                nums[j],nums[j+1]= nums[j+1],nums[j]
    print(nums)
    
            


bubble_sort(nums)
```
