Quick Sort:
1. Pick a pivot
     - it can be the first element
     - it can be the last element
     - it can be middle element
     - it can be any random element
2. Put pivot at its correct position/index

```python
nums =[4,1,7,6,3,2,8]

def partition(nums,low,high):
    p=nums[low]
    i = low
    j = high
    while i<j:
        while nums[i]<=p and i<=high-1:
            i+=1
        while nums[j]>p and j>=low-1:
            j-=1
        if i<j:
            nums[i],nums[j]= nums[j],nums[i]
    nums[low],nums[j] = nums[j],nums[low]
    return j


def quick_sort(nums,low,high):
    if low<high:
       p_index= partition(nums,low,high)
       quick_sort(nums,low,p_index)
       quick_sort(nums,p_index+1,high)
    return nums        
      
    
print(quick_sort(nums,0,len(nums)-1))     
            
            
        
```
