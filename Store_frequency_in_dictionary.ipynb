#### #10 Frequency Map/Dictionary

```python
nums=[5,6,7,7,1,9,111,1,1,5,1,1]
freq_maps= {}
for i in range(0,len(nums)):
    if nums[i] in freq_maps:
        freq_maps[nums[i]]+=1
    else:
        freq_maps[nums[i]] =1

freq_maps
```

```python
# Sorter Way (Smart way)
nums=[5,6,7,7,1,9,111,1,1,5,1,1]
freq_maps= {}
for i in range(0,len(nums)):
    freq_maps[nums[i]] =freq_maps.get(nums[i],0)+1

freq_maps
```
