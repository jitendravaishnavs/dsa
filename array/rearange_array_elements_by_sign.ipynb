```python
nums = [3,1,-2,-5,2,-4]

class Solution:
    def rearrangeArray(self, nums: list[int]) -> list[int]:
        n = len(nums)
        result = [0] * n
        posindex,negindex= 0,1
        for i in range(0,n):
            if nums[i]>=0:
                result[posindex]= nums[i]
                posindex+=2
            else:
                result[negindex]=nums[i]
                negindex+=2
        return result
    
        
        
obj = Solution()
print(obj.rearrangeArray(nums))
```
