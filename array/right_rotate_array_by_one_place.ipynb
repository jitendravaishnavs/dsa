```python
nums = [5,-2,3,9,0,6,10,7]
n=len(nums)
temp = nums[n-1]

for i in range(n-2,-1,-1):
    nums[i+1] = nums[i]
nums[0] = temp
nums
```

```python
## By k places
k=5
nums = [3,9,5,6,7,2]
n= len(nums)
rotations = k%n
for i in range(0,rotations):
    print(i)
    e=nums.pop()
    nums.insert(0,e)
nums
```
